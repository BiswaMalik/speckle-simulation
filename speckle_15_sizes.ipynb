```python
import numpy as np
import matplotlib.pyplot as plt

N = 512                 # for FFT, must be power of 2
WAVELENGTH = 632.8e-9   # HeNe laser, meters
Z = 0.5                 # Propagation distance /distance to observation/camera plane, meters
D_MAX = 5e-3            # physical aperture diameter at radius_fraction = 1.0
RADIUS_FRACTIONS = np.round(np.geomspace(0.03, 0.55, 15), 4)

rng = np.random.default_rng(42)

def circular_aperture_mask(n, radius_fraction):
    """Boolean mask of a centered circular aperture on an n x n grid."""
    y, x = np.indices((n, n)) - n // 2
    r = np.sqrt(x**2 + y**2)
    return r <= (radius_fraction * n / 2)

def speckle_pattern(n, radius_fraction, rng):
    """Fully developed speckle intensity pattern (normalized, <I> = 1)."""
    mask = circular_aperture_mask(n, radius_fraction)
    phase = rng.uniform(0, 2 * np.pi, size=(n, n))
    field = np.exp(1j * phase)
    field[~mask] = 0.0
    E = np.fft.fftshift(np.fft.fft2(np.fft.ifftshift(field)))
    intensity = np.abs(E) ** 2
    return intensity / intensity.mean()

def physical_speckle_size_um(wavelength, D, z):
    """Average speckle diameter in micrometers: d ~ 1.22 * lambda * z / D."""
    return 1.22 * wavelength * z / D * 1e6

def crop(a, n, zoom_frac=0.25):
    c = n // 2
    zm = int(n * zoom_frac)
    return a[c - zm:c + zm, c - zm:c + zm]
```

```python
# Type 1/15: aperture radius fraction = 0.03
D = D_MAX * 0.03                      # illuminated aperture diameter (m)
pattern_1 = speckle_pattern(N, 0.03, rng)
size_um = physical_speckle_size_um(WAVELENGTH, D, Z)

plt.figure(figsize=(4.2, 4.2))
plt.imshow(crop(pattern_1, N), cmap="gray")
plt.title(f"Type 1: D={D*1e3:.2f} mm  |  speckle size \u2248 {size_um:.0f} \u00b5m")
plt.xticks([]); plt.yticks([])
plt.tight_layout()
plt.show()
print(f"Type 1: aperture D={D*1e3:.2f} mm -> speckle size \u2248 {size_um:.1f} um")
```
